```python
import random

print("=================================")
print("🎲      WELCOME LOTTO GAME      🎲")
print("=================================")

while True:

    # Difficulty selection
    print("\nSelect Difficulty Level:")
    print("1. Easy   (1-50, 10 attempts)")
    print("2. Medium (1-100, 7 attempts)")
    print("3. Hard   (1-200, 5 attempts)")

    choice = input("Choose level (1/2/3): ")

    if choice == "1":
        max_number = 50
        max_attempt = 10
    elif choice == "2":
        max_number = 100
        max_attempt = 7
    elif choice == "3":
        max_number = 200
        max_attempt = 5
    else:
        print("❌ Invalid choice. Default set to Medium.")
        max_number = 100
        max_attempt = 7

    # Generate random number
    secret_number = random.randint(1, max_number)

    print(f"\nGuess the number between 1 and {max_number}")
    print(f"You only have {max_attempt} attempts!")

    attempt = 0
    win = False

    # Game loop
    while attempt < max_attempt:

        try:
            guess = int(input("\nEnter your guess: "))
            attempt += 1

            remaining = max_attempt - attempt

            if guess < secret_number:
                print("📉 Too low!")
            elif guess > secret_number:
                print("📈 Too high!")
            else:
                print(f"\n🎉 Congratulations!")
                print(f"You guessed the number in {attempt} attempts.")
                score = (max_attempt - attempt + 1) * 10
                print(f"🏆 Your score: {score}")
                win = True
                break

            if remaining > 0:
                print(f"Remaining attempts: {remaining}")

        except ValueError:
            print("⚠️ Please enter a valid number!")

    # Lose condition
    if not win:
        print("\n💀 Game Over!")
        print(f"The correct number was: {secret_number}")

    # Play again
    play_again = input("\nDo you want to play again? (yes/no): ").lower()

    if play_again != "yes":
        print("\n👋 Thank you for playing Lotto Game!")
        break
```
